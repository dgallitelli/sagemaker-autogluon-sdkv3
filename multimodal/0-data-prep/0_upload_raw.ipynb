# Step 0: Upload Raw Multimodal Churn Data to S3

This notebook copies the raw synthetic churn JSONL data from the SageMaker example bucket to your own S3 bucket using a server-side copy.

## Configuration

Set your region, bucket, and S3 prefix below.

```python
import boto3
import sagemaker

region = boto3.session.Session().region_name
sess = sagemaker.session.Session()
bucket = sess.default_bucket()
s3_prefix = "autogluon-multimodal"
s3_output = f"s3://{bucket}/{s3_prefix}/raw/"

print(f"Region:    {region}")
print(f"S3 output: {s3_output}")
```

## Copy Raw JSONL Files

The source data lives in the SageMaker example files bucket. We copy 3 files: `train.jsonl`, `validation.jsonl`, and `test.jsonl`.

```python
source_bucket = f"sagemaker-example-files-prod-{region}"
source_prefix = "datasets/tabular/synthetic_churn_prediction_with_text"

s3 = boto3.client("s3", region_name=region)

dest_prefix = f"{s3_prefix}/raw"

for filename in ("train.jsonl", "validation.jsonl", "test.jsonl"):
    source_key = f"{source_prefix}/{filename}"
    dest_key = f"{dest_prefix}/{filename}"
    print(f"Copying s3://{source_bucket}/{source_key} -> s3://{bucket}/{dest_key}")
    s3.copy_object(
        CopySource={"Bucket": source_bucket, "Key": source_key},
        Bucket=bucket,
        Key=dest_key,
    )

print(f"\nRaw data uploaded to s3://{bucket}/{dest_prefix}/")
```
